```python
import httpx
from IPython.display import Markdown, display
from a2a.client import (
    Client,
    ClientConfig,
    create_client,
)
from a2a.types import SendMessageRequest
from a2a.helpers.proto_helpers import new_text_message
from a2a.helpers import get_stream_response_text
```

```python
host = "localhost"
port = 9999

prompt = "How much would I pay for mental health therapy?"
```

```python
custom_http = httpx.AsyncClient(timeout=100.0)
config = ClientConfig(httpx_client=custom_http)
client: Client = await create_client(f"http://{host}:{port}", client_config=config)

# agent_card = await client.get_card()
print("Agent Card retrieved!")

message = new_text_message(text=prompt)
request = SendMessageRequest(
    message=message,
)
display(Markdown(f"**Sending prompt:** `{prompt}` to the agent..."))

responses = client.send_message(request)
text_content = ""

# Step 5: Process the responses from the agent
async for response in responses:
    text_content = get_stream_response_text(response)

display(Markdown("### Final Agent Response\n-----"))
if text_content:
    display(Markdown(text_content))
else:
    display(
        Markdown(
            """**No final text content received or task did not 
            complete successfully.**"""
        )
    )
```
